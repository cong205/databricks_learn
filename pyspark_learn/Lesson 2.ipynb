```python
df = (
    spark.read.option("header","True")
    .option("inferSchema", "True")
    .csv("/Volumes/workspace/bronze/source_systems/source_crm/sales_details.csv")
)
```

```python
df.show()
```

# xóa các hàng có giá trị null

```python
df.na.drop().show()
```

## thresh (ngưỡng)

```python
## các hàng vượt quá ngưỡng sẽ bị xóa
## =1 tức có ít nhất 1 not null
df.na.drop(thresh=1).show()
```

## subset (tên cột)

```python
## các hàng có giá trị null trong cột sẽ bị xóa
df.na.drop(subset=['tên cột']).show()
```

# filling the missing value (điền vào các ô còn trống)

```python
## các hàng có giá trị null trong data frame sẽ bị thay thế bằng giá trị 'value'
df.na.fill('value').show()
```

```python
## các hàng có giá trị null trong cột sẽ bị thay thế bằng giá trị 'value'
df.na.fill('value','tên cột').show()
df.na.fill('value',['tên cột 1','tên cột 2','tên cột n']).show()
```

## có thể thay bằng mean (trung bình), median (trung vị) của các values trong cột chứa nó

# filter operations (các thao tác lọc data)

```python
df.filter("sls_price < 3400").show()
```

```python
## lấy các hàng có giá < 3400 của 2 cột sls_prd_key và sls_cust_id
df.filter("sls_price < 3400").select(['sls_prd_key','sls_cust_id']).show() 
```

```python
df.filter((df['sls_price'] < 3400)).select(['sls_prd_key','sls_cust_id']).show() 
```

```python
df.filter((df['sls_price'] < 3400) & (df['sls_price']  > 1000)).select(['sls_prd_key','sls_cust_id']).show() 
```

```python
## nghịch đảo của filter (thêm dấu ~ trước điều kiện)
df.filter(~((df['sls_price'] < 3400))).select(['sls_prd_key','sls_cust_id']).show()
# lấy các hàng có giá > 3400 của 2 cột sls_prd_key và sls_cust_id
```

# group by

```python
# các hàm khác: mean(), count(), max(), min(), avg()
df.groupBy('sls_cust_id').sum('sls_price').show()
```

```python
# tính tổng gía trị của cột sls_price
df.agg({'sls_price':'sum'}).show()
```
